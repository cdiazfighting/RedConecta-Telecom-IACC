Proyecto de Ciencia de Datos - Semana 8 IACC

Objetivo: analizar latencia, incidencias y SLA.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_excel('dataset_set_D_telecomunicaciones.xlsx')
print('Registros:', len(df))
display(df.head())
```

```python
print('Incidencias por región:')
display(df.groupby('region')['incidencias_reportadas'].sum())
print('Tiempo promedio de resolución:')
display(df.groupby('tipo_servicio')['tiempo_resolucion_hr'].mean())
print('Cumplimiento SLA:')
display(df['cumplimiento_sla'].value_counts())
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
df.groupby('region')['latencia_ms'].mean().plot(kind='bar', ax=ax[0], title='Latencia por región')
df['cumplimiento_sla'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=ax[1], title='Cumplimiento SLA')
ax[0].set_ylabel('Latencia (ms)')
ax[1].set_ylabel('')
plt.tight_layout()
plt.show()
```
